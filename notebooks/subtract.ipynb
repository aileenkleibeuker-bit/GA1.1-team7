```python
def substract(a, b):
    return a - b

print(subtract(10, 4))    # 6
print(subtract(5, 5))     # 0
print(subtract(3, 8))     # -5
```

```python
def distance_from_zero(x):
    if x < 0:
        return substract(0, x)
    else:
        return x

print(distance_from_zero(-5))  # 5
print(distance_from_zero(3))   # 3
print(distance_from_zero(0))   # 0
```
